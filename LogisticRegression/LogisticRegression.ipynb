# Logistic regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("../datasets/Iris.csv")
df
```

```python
df = df.iloc[:100 , :]
df = df.drop("Id", axis=1)
df["Species"] = df["Species"].apply(lambda x: 0 if x=="Iris-setosa" else 1)
df
```

```python
y_train = df["Species"]
x_train = df.drop("Species", axis=1)

y_train = np.array(y_train).reshape((-1, 1))
x_train = np.array(x_train)
x_train
```

Сигмоида:

```python
def g(z): 
    z = np.clip(z, -500, 500)
    response = 1 / (1 + np.exp(-z))
    return response
```

```python
len(x_train)
```

### Градиент функции потерь 

```python
size_of_features = x_train.shape[1]
np.random.seed(21)
w = np.random.randn(size_of_features).reshape((size_of_features,-1)) #начальные веса
alpha = 0.1 

w
```

```python
z = 0
iterator = 0
loss = np.zeros(size_of_features).reshape((size_of_features,-1))
prev_loss = np.zeros(size_of_features).reshape((size_of_features,-1))
while True:
    loss = np.zeros(size_of_features).reshape((size_of_features,-1))
    for i in range(len(x_train)):  # цикл подсчета градиента от loss (идем по каждой строке датафрейма (100 операций))
        z=0
        x = x_train[i]
        for j in range(size_of_features):
            z += x[j] * w[j]                # z = w1*x1 + w2*x2 + w3*x3 ...
        err = (y_train[i] - g(z))
        for j in range(size_of_features):
            loss[j] -= x[j] * err           # Подсчет градиента от функции потерь

    iterator +=1
    #if (abs(prev_loss) - abs(loss)).all() < 1e-4:
    if np.linalg.norm(loss) < 1e-4:
    #if iterator == 100:
        break
    w = w - np.dot(alpha, loss)
    prev_loss = loss
loss, iterator
```

```python
predict = np.zeros(len(y_train)).reshape((-1, 1))
list_of_predict = list()
z_predicted = x_train.dot(w)
for i in range(len(z_predicted)):
    predict[i] = g(z_predicted[i])
    if predict[i] < 0.5:
        predict[i] = 0
    else:
        predict[i] = 1
    list_of_predict.append(predict[i])

list_of_predict = [x.item() for x in list_of_predict]
df_predict = pd.Series(list_of_predict).astype('Int64')
df_predict.value_counts()
```

```python
comparison = pd.DataFrame({"Actual data": df["Species"], "NUMPY predict": df_predict})
comparison.head(40)
```

## Сверим работу модели с работой LogisticRegression из sklearn

```python
clf = LogisticRegression()
clf.fit(x_train, y_train)
```

```python
sklearn_predict = clf.predict(x_train)
comparison['sklearn predict'] = sklearn_predict
```
